```python
from juliacall import Main as jl
from juliacall import Pkg

def bar(p, age):
    p.age = age

def foo(p, age):
    bar(p, age)

def set_age_py(p, age):
    foo(p, age)
    
jl.seval("""
mutable struct Person 
    name::String
    age::Int
end

module PersonModule

using .Main: Person

function set_age!(p::Person, age::Int)
    println("Person before: ", p)
    PersonModule.set_age_py(p, age)
    println("Person after: ", p)
end

end

using .PersonModule
""")

jl.PersonModule.set_age_py = set_age_py

jl.seval("""
PersonModule.set_age!(Person(\"Alice\", 30), 35)         
""")
```
